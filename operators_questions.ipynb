```python
"""Consider a room having one door and two windows both of the same size. 
Accept dimensions of the room, door and window. Print the area to be 
painted (interior walls) and area to be whitewashed (roof)."""

# dimensions of room
H = int(input("enter hight of room:"))
L = int(input("enter length of room"))
W = int(input("enter width of room"))

# dimensions of door
l = int(input("enter lenght of door"))
h = int(input("enter hight of door"))

# dimensions of windows
l1 = int(input("enter lenght of W"))
h1 = int(input("enter hight of W"))

# area of interior walls
area = 2*H * (L+W)
print(area)

# area of window and door
area_door = l * h
area_w = l1 * h1
open_area = area_door - 2 * area_w
print(open_area)

# paint area

paint_area = area - open_area
print(paint_area)

# roof area

roof_area = L * W
print(roof_area)
```

```python
"""A timestamp is three numbers: a number of hours, minutes and seconds. 
Given two timestamps, calculate how many seconds is between them. 
The moment of the first timestamp occurred before the moment of the second timestamp"""

h1 = int(input("enter the hours1: "))
m1 = int(input("enter the minute1: "))
s1 = int(input("enter the seconds1: "))

h2 = int(input("enter the hours2:"))
m2 = int(input("enter the minutes2:"))
s2 = int(input("enter the seconds2: "))

time1 = h1 * 3600 + m1 * 60 + s1
time2 = h2 * 3600 + m2 * 60 + s2

difference = (time2 - time1)
print(difference)
```

```python
"""A school decided to replace the desks in three classrooms. Each desk sits two students. 
Given the number of students in each class, print the smallest possible number of 
desks that can be purchased. The program should read three integers: 
the number of students in each of the three classes, a, b and c respectively. 
eg: 
The first class has 20 students and thus needs 10 desks. 
The second class has 21 students, so they can get by with no fewer than 11 desks.
11 desks is also enough for the third class of 22 students. So we need 32 desks in total."""


a = int(input("enter the number of students class 1 :"))
b = int(input("enter the number of students class 2 :"))
c = int(input("enter the number of students class 3 :"))

class1 = (a + 1) // 2
class2 = (b + 1) // 2
class3 = (c + 1) // 2

smallest_value =  class1 + class2 + class3

print(smallest_value)
```
