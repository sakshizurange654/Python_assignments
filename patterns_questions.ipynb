```python
''' 30. Write a python program to read a number and print a right triangle using "*"
Eg : 
Input : 5
----------Output--------- 
*
* *
* * *
* * * *
* * * * * '''

num = int(input("enter the number: "))
for i in range(1, num+1):
    print("*" *i)
```
