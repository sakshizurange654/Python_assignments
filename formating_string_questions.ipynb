```python
"""5. Write a Python program to print the number of prime numbers which are less than or equal to an given integer. 
Input: Input the number(n):35
Number of prime numbers which are less than or equal to n.: 11"""

n = int(input("enter the number(n):"))
count = 0

for i in range(2, n + 1):
    for num in range(2, i):
        if i % num == 0:
            break
    else:
        count += 1
        
print("no. of prime numbers which are less than or equal to n.:", count)
```

```python
"""7. Write a Python program to find the longest common prefix string amongst an given array of strings. Return false If there is no common prefix.
For Example, longest common prefix of "abcdefgh" and "abcefgh" is "abc". 
Sample Output:abc"""

strings = ["abcdefgh", "abcefgh"]

prefix = strings[0] if strings else ""
for s in strings[1:]:
    while not s.startswith(prefix):
        prefix = prefix[:-1]
        if not prefix:
            break

print(prefix if prefix else False)
        
```

```python
# Write a python program to convert alternate characters to capital letters

a = "abcdefghijklmnopqrstuvwxyz"
string = ""
for i in range (len(a)):
    if i % 2 == 0:
        string += a[i].lower()
    else:
        string += a[i].upper()
print(string)
```

```python
"""Given a string input Count all lower case, upper case, digits, and special symbols
eg :    Total counts of chars, digits,and symbols Chars = 8 Digits = 3 Symbol = 4"""

a = "SakshI@SaNTosh#654"

chr_count = 0
digit_count = 0
symbol_count = 0

for i in a:
    if i.isalpha():
        chr_count += 1
    elif i.isdigit():
        digit_count += 1
    else:
        symbol_count += 1
        
print(chr_count)
print(digit_count)
print(symbol_count)
```
