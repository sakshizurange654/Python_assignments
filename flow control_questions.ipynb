```python
"""WAP to accept any year from user and check if its a leap year or not 
A leap year is a year which is divisible by 4 but not by 100 
    and should be divisible by 400 """

year = int(input("enter the year:"))
i = 1
while i <= 1:
    if (year % 4 == 0 or year % 100 != 0)  and (year % 400 == 0):
        print("its a leap year:")
    else:
        print("its not leap year:")
        
    i += 1
```

```python
""" WAP to accept Marks of 5 subjects and print the class obtained 
    The classes are as follows :
    1) Distinction if % is >= 70 
    2) 1st Class if % is between 60 and 69
    3) 2nd class if % is between 50 and 59 
    4) Fail if % is less than 50 """

sub1 = int(input("enter the marks:"))
sub2 = int(input("enter the marks:"))
sub3 = int(input("enter the marks:"))
sub4 = int(input("enter the marks:"))
sub5 = int(input("enter the marks:"))

total = sub1+sub2+sub3+sub4+sub5 
avg = total/5

if avg >= 70:
    print("your are in distinction")
elif (avg >= 60 and avg <= 69):
    print("1st class")
elif (avg >= 50 and avg <= 59):
    print("2nd class")
elif avg < 50:
    print("you are fail")
else:
    print("try for next year")
```
