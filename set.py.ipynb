```python
#1.] Convert a list with duplicates into a set and print all unique values.

listt = [1,2,3,4,5,6,3,2,4,7,8,9,1]

unique_values = set(listt)
print(unique_values)
```

```python
#2.] Given two sets of numbers, find the common elements 
#(intersection).

set_1 = {1,2,5,7}
set_2 = {0,2,5,9}

print(set_1 & set_2)
```

```python
# 3.] Given two sets, find the elements that are only in the first set.

set_1 = {1,2,5,7}
set_2 = {2,5,7}

print(set_1 - set_2)
```

```python
# 4.] Given two sets, find the elements that are in either set but not both 
# (symmetric difference).

set_1 = {1,2,5,7}
set_2 = {5,7,3,4}

a = set_1.symmetric_difference(set_2)
print(a)
```

```python
# 5.] Check if one set is a subset of another set.

sett = {4,8,12,16,20}
sset = {4,12}

check = sset.issubset(sett)
print(check)
```

```python
#  6.] Add a new element to a set of fruits.

fruits = {"watermelon", "apple", "orange", "Mango"}
print(fruits)

fruits.add("papaya")
print(fruits)
```

```python
#  7.] Remove an element from a set only if it exists (use discard).

fruits = {"watermelon", "apple", "orange", "Mango"}
print(fruits)
fruits.discard("apple")
print(fruits)
```

```python
# 8.] Create two sets of colors and find the combined unique colors using union.

set_1 = {"red", "pink"}
set_2 = {"red", "white"}

print(set_1.union(set_2))
```

```python
# 9.] From a list of names, find how many unique names exist.

list_1 = ["sakshi","sayali","rushi","sakshi","ketan","sayali","rushi"]

new = set(list_1)
print(new)
print(len(new))
```

```python
# 10.] Given a set of numbers, check if a number exists inside the set (membership test).

a = {4,8,12,16,20}
print(a)
print(2.in(a))
```

```python
x = {"red", "white"}

print(x)
print("yellow" in(x))
print("yellow" not in(x))
```

```python
# 11.] From a sentence, extract all unique words using a set.

sentence = "sakshi santosh zurange"
new = sentence.split()
print(set(new))
```

```python
# 12.] Given a set of even numbers, check if {2, 4} is a subset.

a = {2,4,6,8,10}
b = {2,4}
print(b.issubset(a))
```

```python
#13.] Convert a string "mississippi" into a set and count how many unique characters it has.

string = "mississippi"
a = set(string)
print(a)
print(len(a))
```

```python
# 14.] Given two sets of students, find the students who are in either class A or class B.


set_1 = {"sakshi", "ram", "shyam", "vidya"}
set_2 = {"ram", "sayali", "bablu", "sakshi"}

print(set_1 & set_2)
```

```python
# 15.] Given a set, remove a random element using pop().

set_1 = {"sakshi", "ram", "shyam", "vidya"}
a = set_1.pop()
print(a)
print(set_1)
```

```python
# 17.] Given two sets, check if they are disjoint (no common elements).

set_1 = {"red", "pink"}
set_2 = {"black", "white"}

print(set_1.isdisjoint(set_2))
```

```python
# 18.] Add multiple elements to a set at once using update().

a = {2,4,6,8,10}
a.update([12,14])
print(a)
```
